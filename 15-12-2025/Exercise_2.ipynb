```python
csv_df=spark.read.option("header",True).option("inferSchema",True).csv("trips_csv")
csv_df.filter(
    (col("trip_fare")>400)&(col("trip_status")=="Completed")

).show()
```

```python
csv_df.select("trip_id","city","vehicle_type","trip_fare")\
  .orderBy(col("trip_fare").desc())\
  .show()
```
