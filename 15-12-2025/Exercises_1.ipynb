```python
orders_csv_df=spark.read\
  .option("header",True)\
  .option("inferSchema",True)\
  .csv("orders_csv")
filtered_df=orders_csv_df.filter("order_amount>700")
filtered_df.show()
```

```python
delivered_df=df.filter(df.order_status=="Delivered")
delivered_df.write.mode("overwrite")\
  .json("delivered_orders_json")
```

```python
json_df=spark.read.json("delivered_orders_json")
json_df.filter(
    (json_df.city=="Mumbai")&
    (json_df.payment_mode=="Card")
).show()
```

```python
from pyspark.sql.functions import when
category_df=json_df.withColumn(
    "delivery_category",
    when(json_df.delivery_time_minutes<=30,"Fast Delivery")
    .when(json_df.delivery_time_minutes<=45,"Medium Delivery")
    .otherwise("Slow Delivery")
)
category_df.show()
```

```python
category_df.coalesce(1)\
  .write.mode("overwrite")\
  .json("single_partition_json")
```

```python
from pyspark.sql.functions import sum
df.groupBy("cuisine")\
  .agg(sum("order_amount").alias("total_sales"))\
  .orderBy("total_sales",ascending=False)\
  .show(1)
```

```python
df.groupBy("city")\
  .count()\
  .orderBy("count",ascending=False)\
  .show(1)
```
